# Practice Assignment: Hypothesis Testing

**Topics:** One-sample z-test, one-sample t-test, and Welch's two-sample t-test

**Instructions:** Each question has five parts, (a)–(e), worth 4 points each. For every part, determine whether the statement is **TRUE or FALSE** and justify your answer.

A statement is TRUE only if every claim and its reasoning are correct. Show key calculation steps. Numerical answers may be rounded to three or four decimal places.

## Question 1 (20 points). One-sample z-test

A laboratory monitors the concentration of a chemical in a water supply. Historical measurements indicate that the population standard deviation is known to be $\sigma=4$ mg/L. A random sample of $n=64$ independent measurements has sample mean $\bar{x}=21$ mg/L. The historical population mean is $\mu_0=20$ mg/L.

The laboratory wants to test whether the current mean concentration is greater than the historical mean, using significance level $\alpha=0.05$.

Determine whether each of the following statements is TRUE or FALSE. Justify your answers.

### (a)

An appropriate pair of hypotheses is

$
H_0:\mu=20, \qquad H_1:\mu>20.
$

### (b)

Under the null hypothesis, the test statistic

$
Z=\frac{\bar{X}-20}{4/\sqrt{64}}
$

follows a standard normal distribution only if the population measurements are Gaussian.

### (c)

The observed test statistic is $Z=2$, and the one-sided p-value is approximately $0.0228$.

### (d)

If the researcher had instead specified the two-sided alternative $H_1:\mu\neq20$, the p-value for the same observed statistic would be approximately $0.0456$, and the null hypothesis would still be rejected at $\alpha=0.05$.

### (e)

If the sample mean had instead been 19 mg/L, the one-sided p-value for the alternative $H_1:\mu>20$ would be smaller than $0.05$, because the observed mean is one standard error away from the null mean.

## Question 2 (20 points). One-sample t-test

A manufacturing process is designed to produce components with a mean diameter of 50 mm. A researcher collects $n=16$ independent observations and obtains a sample mean of $\bar{x}=52$ mm and a sample standard deviation of $s=4$ mm. Assume the population distribution is Gaussian.

The researcher wants to test whether the population mean diameter differs from 50 mm, using significance level $\alpha=0.05$.

Determine whether each of the following statements is TRUE or FALSE. Justify your answers.

### (a)

Since the population standard deviation is unknown, an appropriate test statistic is

$
T=\frac{\bar{X}-50}{S/\sqrt{16}}.
$

### (b)

Under the null hypothesis, the test statistic follows a Student's t-distribution with 15 degrees of freedom.

### (c)

The observed test statistic is $T=2$, and the two-sided p-value is approximately $0.032$, so the null hypothesis is rejected at significance level $\alpha = 0.05$.

### (d)

If the population standard deviation were known to be exactly 4 mm instead of estimated by the sample, the observed standardized statistic would still equal 2, but its reference distribution would be standard normal rather than $t_{15}$.

### (e)

The t-distribution has heavier tails than the standard normal distribution because using the sample standard deviation introduces additional uncertainty into the test statistic.

## Question 3 (20 points). Welch's two-sample t-test

A university compares the final examination scores of students taught using two different instructional methods. The two groups consist of independent random samples. The population variances are not assumed to be equal.

The sample summaries are:

|  | Method A | Method B |
|---|---:|---:|
| Sample size | 10 | 20 |
| Sample mean | 78 | 72 |
| Sample SD | 8 | 4 |

The researcher wants to test whether the two population means differ, using significance level $\alpha=0.05$.

Determine whether each of the following statements is TRUE or FALSE. Justify your answers.

### (a)

An appropriate pair of hypotheses is

$
H_0:\mu_A-\mu_B=0, \qquad
H_1:\mu_A-\mu_B\neq0.
$

### (b)

The estimated standard error used in Welch's test is

$
SE=\sqrt{\frac{8^2}{10}+\frac{4^2}{20}}
=\sqrt{7.2}.
$

### (c)

The observed Welch test statistic is approximately $t=2.24$.

### (d)

The degrees of freedom for Welch's test are necessarily equal to

$
10+20-2=28,
$

because the two samples are independent.

### (e)

The Welch test uses a t-distribution with approximate degrees of freedom because its standard error combines two separately estimated sample variances, and the resulting statistic does not generally have an exact Student's t-distribution with $n_A+n_B-2$ degrees of freedom.
