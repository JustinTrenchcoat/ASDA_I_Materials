## Question 1 (20 points). One-sample z-test

A laboratory monitors the concentration of a chemical in a water supply. Historical measurements indicate that the population variance is known to be $\sigma^2=16$ mg/L. A random sample of $n=64$ independent measurements has sample mean $\bar{x}=21$ mg/L. The historical population mean is $\mu_0=20$ mg/L.

The laboratory wants to test whether the current mean concentration is greater than the historical mean, using significance level $\alpha=0.05$.

Determine whether each of the following statements is TRUE or FALSE. Justify your answers.

### (a)

An appropriate pair of hypotheses is

$
H_0:\mu=20, \qquad H_1:\mu>20.
$

### (b)

Under the null hypothesis, the test statistic

$
Z=\frac{\bar{X}-20}{4/\sqrt{64}}
$

follows a standard normal distribution if the population measurements are Gaussian.

### (c)

The observed test statistic is $Z=2$, and our p-value is approximately $0.9772$.

### (d)

If the researcher had instead specified the two-sided alternative $H_1:\mu\neq20$, the p-value for the same observed statistic would be approximately $0.0456$, and the null hypothesis is true at $\alpha=0.05$. What about $\alpha=0.01$?

### (e)

If the sample mean had instead been 19 mg/L, the one-sided p-value for the alternative $H_1:\mu>20$ would be smaller than $0.05$, because the observed mean is one standard error away from the null mean.

## Question 2 (20 points). One-sample t-test

A manufacturing process is designed to produce components with a mean diameter of 50 mm. A researcher collects $n=16$ independent observations and obtains a sample mean of $\bar{x}=52$ mm and a sample variance of $4$ mm. Assume the population distribution is Gaussian.

The researcher wants to test whether the population mean diameter differs from 50 mm, using significance level $\alpha=0.05$.

Determine whether each of the following statements is TRUE or FALSE. Justify your answers.

### (a)

Since the population standard deviation is unknown, an appropriate test statistic is

$
T=\frac{\bar{X}-50}{4/\sqrt{16}}.
$

### (b)

Under the null hypothesis, the test statistic follows a Student's t-distribution with 15 degrees of freedom.

### (c)

The observed test statistic is $T=4$, and the two-sided p-value is approximately between $0.005$ and $0.001$, so we accept the alternative hypothesis at significance level $\alpha = 0.05$.

### (d)

If the population variance were known to be exactly 4 mm instead of estimated by the sample, the observed test statistic would still equal 4, but its distribution would be standard normal rather than $t_{15}$.

### (e)

The t-distribution has heavier tails than the standard normal distribution because using the sample standard deviation introduces additional uncertainty into the test statistic.

The observed Welch test statistic is approximately $t=2.24$.

### (d)

The degrees of freedom for Welch's test is approximately 11. Under significance level $\alpha=0.05$, we reject the null hypothesis.
